# Diabetes Patient Readmission Prediction

## Data Analysis and Preprocessing

### Objective
The objective of this project is to analyze the Diabetes 130-US Hospitals dataset, understand its structure, identify missing values and data quality issues, preprocess the dataset, and prepare it for machine learning models that predict hospital readmission.

```python
# Import libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder

# Display plots inside notebook
%matplotlib inline

# Ignore warnings
import warnings
warnings.filterwarnings("ignore")
```

```python
# Load dataset

df = pd.read_csv("../data/diabetic_data.csv")

# Display first five rows
df.head()
```

```python
df.describe(include='object')
```

```python
df.replace("?", pd.NA, inplace=True)
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
# Drop columns with too many missing values
df.drop(columns=["weight", "max_glu_serum", "A1Cresult", "medical_specialty", "payer_code"], inplace=True)

# Fill missing values in remaining columns
df["race"] = df["race"].fillna(df["race"].mode()[0])
df["diag_1"] = df["diag_1"].fillna(df["diag_1"].mode()[0])
df["diag_2"] = df["diag_2"].fillna(df["diag_2"].mode()[0])
df["diag_3"] = df["diag_3"].fillna(df["diag_3"].mode()[0])
```

```python
df.isnull().sum()
```

```python
import os

os.makedirs("../cleaned_data", exist_ok=True)

df.to_csv("../cleaned_data/cleaned_diabetic_data.csv", index=False)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in df.select_dtypes(include='object').columns:
    df[col] = le.fit_transform(df[col].astype(str))

df.head()
```

```python
pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, pred))
print("Precision:", precision_score(y_test, pred, average="weighted"))
print("Recall:", recall_score(y_test, pred, average="weighted"))
print("F1 Score:", f1_score(y_test, pred, average="weighted"))
```
