```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/raw/diabetic_data.csv")

print(df.shape)
df.head()
```

```python
df.replace("?", np.nan, inplace=True)
```

```python
df.isnull().sum()
```

```python
missing_percentage = (df.isnull().sum()/len(df))*100

missing_percentage.sort_values(ascending=False)
```

```python
missing_df = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": missing_percentage
})

missing_df = missing_df[missing_df["Missing Values"] > 0]

missing_df.sort_values("Percentage", ascending=False)
```

```python
missing_df.loc[["weight","payer_code","medical_specialty"]]
```

```python
df.drop(columns=["weight"], inplace=True)
```

```python
df["race"] = df["race"].fillna(df["race"].mode()[0])
```

```python
df["medical_specialty"] = df["medical_specialty"].fillna("Unknown")
```

```python
df["payer_code"] = df["payer_code"].fillna("Unknown")
```

```python
print(df.duplicated().sum())
```

```python
df.drop(columns=["encounter_id","patient_nbr"], inplace=True)
```

```python
print(df.shape)
```

```python
df["readmitted"].value_counts()
```

```python
df["readmitted"] = df["readmitted"].replace({
    "<30":1,
    ">30":0,
    "NO":0
})
```

```python
df["readmitted"].value_counts()
```

```python
for col in df.columns:
    print(f"{col}: {df[col].nunique()}")
```

```python
constant_columns = [col for col in df.columns if df[col].nunique()==1]

constant_columns
```

```python
df.drop(columns=constant_columns, inplace=True)
```

```python
df.info()
```

```python
df.to_csv("../data/processed/diabetes_cleaned.csv", index=False)
```

```python
print("Cleaned dataset saved successfully!")
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    print("=" * 50)
    print(col)
    print(df[col].value_counts().head(10))
```

```python
age_map = {
    "[0-10)":5,
    "[10-20)":15,
    "[20-30)":25,
    "[30-40)":35,
    "[40-50)":45,
    "[50-60)":55,
    "[60-70)":65,
    "[70-80)":75,
    "[80-90)":85,
    "[90-100)":95
}

df["age"] = df["age"].map(age_map)
```

```python
df["gender"].value_counts()
```

```python
df = df[df["gender"] != "Unknown/Invalid"]
```

```python
medication_map = {
    "No":0,
    "Steady":1,
    "Up":2,
    "Down":3
}

df["insulin"] = df["insulin"].map(medication_map)

```

```python
binary_map = {
    "No":0,
    "Ch":1
}

df["change"] = df["change"].replace(binary_map)

df["diabetesMed"] = df["diabetesMed"].replace({
    "No":0,
    "Yes":1
})
```

```python
a1c_map = {
    "None":0,
    "Norm":1,
    ">7":2,
    ">8":3
}

df["A1Cresult"] = df["A1Cresult"].replace(a1c_map)
```

```python
glu_map = {
    "None":0,
    "Norm":1,
    ">200":2,
    ">300":3
}

df["max_glu_serum"] = df["max_glu_serum"].replace(glu_map)
```

```python
df = pd.get_dummies(
    df,
    columns=["race"],
    drop_first=True
)
```

```python
df["total_visits"] = (
    df["number_outpatient"]
    + df["number_emergency"]
    + df["number_inpatient"]
)
```

```python
df["total_procedures"] = (
    df["num_lab_procedures"]
    + df["num_procedures"]
)
```

```python
import matplotlib.pyplot as plt

df["readmitted"].value_counts().plot(kind="bar")

plt.title("Readmission Distribution")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=["int64","float64"])

corr = numeric_df.corr()

corr["readmitted"].sort_values(ascending=False)
```
